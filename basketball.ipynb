```python
df['points_plus_assists'] = df['points'] + df['assists']
df['points_per_game'] = df['points'] / (df['games_played'] / 36)
```

```python
print (df.head())
```

```python
conditions=[
    (df['position'] == 'PG'),
    (df['position'] == 'SG'), 
    (df['position'] == 'SF'),
    (df['position'] == 'PF'),
    (df['position'] == 'C')
]
choices = ['Point Guard', 'Shooting Guard', 'Small Forward', 'Power Forward', 'Center']
df['position_full'] = np.select(conditions, choices, default='Unknown')

print(df[['name', 'position', 'position_full']].head())
```

```python
df.to_csv('players_updated.csv', index=False)
print("Updated CSV file saved as 'players_updated.csv'")
```
